```python
import numpy as np
import faiss
from app.rag import _embed_model  # reuse the same model instance rag.py already loaded

query_vec = _embed_model.encode(["What company is this person interning at?"])
query_vec = np.array(query_vec, dtype="float32")
faiss.normalize_L2(query_vec)

scores, indices = index.search(query_vec, len(chunks))
for rank, (score, idx) in enumerate(zip(scores[0], indices[0])):
    if "greateway" in chunks[idx].lower():
        print(f"RANK {rank+1} of {len(chunks)}, score={score:.3f}")
        print(chunks[idx])
```
